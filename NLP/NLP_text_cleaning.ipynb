# Text clean

* Lowercasing -> "Hello WORLD" → "hello world"
* Remove Punctuation -> "Hello!!!" → "Hello"
* Remove Numbers -> "I have 2 cats" → "I have cats"
* Remove Extra Whitespaces -> "Hello    World" → "Hello World"
* Remove HTML Tags -> "<>Hello</>" → "Hello"
* Remove URLs -> "Visit https://example.com" → "Visit"
* Remove Email Addresses -> "abc@gmail.com" → ""
* Remove Emojis -> "I ❤️ NLP 😊" → "I NLP"
* Remove Special Characters -> "@Hello#$" → "Hello"
* Expand Contractions -> "don't" → "do not"
* Remove Stopwords -> "The cat is sleeping" → "cat sleeping"
* Tokenization -> "I love NLP" -> ["I", "love", "NLP"]
* Stemming -> playing → play studies → studi running → run
* Lemmatization -> running → run better → good studies → study
* Spelling Correction (Optional) -> "machne learnng" → "machine learning"
* Remove Duplicate Words (Optional) -> "good good movie" → "good movie"
* Normalize Text -> "u" → "you" "gr8" → "great"

#### 1. Lowercasing

```python
lower_fun = (lambda x:x.lower())
```

```python
df["text"] = df["text"].apply(lower_fun)
```

#### 2. Remove Punctuation (. , ! ? : ;)

```python
import string 

def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
df["text"] = df["text"].apply(remove_punc)
```

### 3. Remove Numbers

```python
import re

# text = "I have 2 cats and 1 dog." 
# # re.sub(pattern, replacement, string)
# re.sub(r"\d+", "", text)

def remove_num(txt):
    return re.sub(r"\d+", "", txt)
```

```python
df["text"] = df["text"].apply(remove_num)
```

### 4. Remove Extra Spaces

```python
import re

text = "Hello      World      NLP   "
print(re.sub(r"\s+", " ", text).strip())

def remove_space(txt): 
    return re.sub(r"\s+", " ", txt)
```

```python
df["text"] = df["text"].apply(remove_space)
```

### 5. Remove HTML Tags

```python
from bs4 import BeautifulSoup

text = "<h1>Hello</h1><p>Welcome to NLP</p>" 
BeautifulSoup(text, "html.parser").get_text()
```

### 6. Remove URLs

```python
import re

text = "Visit https://google.com today."
# Jo bhi text http se start ho ya www se start ho aur next space tak chale, usko match karo.
# \s → whitespace (space, tab, newline)
# \S → NOT whitespace

remove_space(re.sub(r"http\S+|www\S+", "", text))
```

### 7. Remove Email Addresses

```python
import re

text = "Contact me at abc@gmail.com" 
re.sub(r"\S+@\S+|.S+", "", text)
```

### 8. Remove Emojis

```python
import emoji

text = "I ❤️ Python 😊" 
emoji.replace_emoji(text, "")
```

### 9. Remove Special Characters

```python
import re

text = "@Hello #Python $100 %Machine^"
re.sub(r"[^a-zA-Z0-9\s]", "", text)
```

### 10. Expand Contractions 

```python
import contractions 

text = "I'm learning NLP. Don't worry."
contractions.fix(text)
```

```python
def expand_contraction(txt): 
    return contractions.fix(txt)
```

```python
df["text"] = df["text"].apply(expand_contraction)
```

```python
df["text"].iloc[0]
```

### 11. Tokenization

```python
from nltk.tokenize import word_tokenize

import nltk
nltk.download('punkt_tab')

text = "I love Machine Learning."
word_tokenize(text)
```

```python
def tokenizing(txt):
    return word_tokenize(txt)
```

```python
df["text"] = df["text"].apply(tokenizing)
```

```python
df["text"].iloc[0]
```

### 12. Remove Stopwords

```python
from nltk.corpus import stopwords
stopwords = set(stopwords.words("english"))

def remove_stopwords(txt):
    return " ".join([word for word in txt if word.lower() not in stopwords])
```

```python
df["text"] = df["text"].apply(remove_stopwords)
```

```python
df["text"].iloc[0]
```

# Bag Of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love Python",
    "I love Machine Learning",
    "Python is amazing"
] 

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(documents)

print("Vocabulary : ")
print(vectorizer.vocabulary_)
print()
print("Features name : ")
print(vectorizer.get_feature_names_out())
print()
print("Bag of Words Matrix:")
print(bow.toarray())
```
