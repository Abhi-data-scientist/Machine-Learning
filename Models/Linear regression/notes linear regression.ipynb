```python
# Linear regression =>
# it is a supervised ml algorithm
# it is used to predict a countinous numerical value
# by finding the relationship b/w ip feature X and o/p feature Y
# Example ->
# predicting salary based on experience
# predicting house price based on area
```

```python
# working process of linear regression model =>
# step 1 -> take input feature x and output features y
# step 2 -> Initialize random weights and bayes
# step 3 -> Now apply linear regression equation
#         y = m1x1 + m2x2 + ... mnxn + b
# step 4 -> now we will predict the output value
# step 5 -> now comapre actual and predicted values
# step 6 -> now calculate the loss function
# step 7 -> now adjust weights and bayes to reduce the error
# step 8 -> repeat the predication and error calculation multiple times
# step 9 -> now we will find the best fit line with minimum error
# step 10 -> now we will use final trend model for predication on unseen or new data
```

```python
# MAE(Mean absolute error)
# Matrix ->
# When to use -> When all errors should be treated equally
# why we use it -> it gives avg error without havily analyzing last mistakes
# ,

# MSE(Mean squared error)
# Matrix
# When to use -> When large errors are very dangerous and importent
# why we use it -> squaring increases the impact large errors
# ,

# RMSE(Root mean squared error)
# Matrix
# When to use -> When you want error in the same unit as
#   target variable and also want to punish large errors
# why we use it -> easy to interpret and sensetive to outliers
# ,

# R2 score
# When to use -> when you want to know how well the model explains the data
# why we use it -> it shows the overall model percentage
```
