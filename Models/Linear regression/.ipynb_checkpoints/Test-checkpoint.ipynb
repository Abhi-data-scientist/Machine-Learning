```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.preprocessing import LabelEncoder 
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('tips - tips.csv')
df.head(3)
```

```python
le = LabelEncoder()
```

```python
df['sex'] = le.fit_transform(df['sex'])
df['smoker'] = le.fit_transform(df['smoker'])
df['day'] = le.fit_transform(df['day'])
df['time'] = le.fit_transform(df['time'])
```

```python
df.head(3)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop(columns=['total_bill']), df['total_bill'], test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
y_test.values
```

```python
lr_mse = mean_squared_error(y_test, y_pred)
lr_r2 = r2_score(y_test, y_pred)
print('MSE: ', lr_mse)
print('R2 score: ', lr_r2)
print('cofficients: ', lr.coef_)
```

# Lasso

```python
from sklearn.linear_model import LassoCV
lasso_cv = LassoCV(
    alphas = [0.001,0.01,0.1,1,10],
    cv = 5,
    random_state = 42
)
lasso_cv.fit(x_train, y_train)
print('Alpha : ', lasso_cv.alpha_)
```

```python
lasso_model = Lasso(alpha=0.001)
lasso_model.fit(x_train,y_train)
```

```python
lasso_pred = lasso_model.predict(x_test)
```

```python
lasso_pred
```

```python
lasso_mse = mean_squared_error(y_test, lasso_pred)
lasso_r2 = r2_score(y_test, lasso_pred)

print('MSE: ', lasso_mse)
print('R2 score: ', lasso_r2)
print('cofficients: ', lasso_model.coef_)
```

# Ridge

```python
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(
    alphas = [0.001,0.01,0.1,1,10],
    cv = 5
    # random_state = 42
)
ridge_cv.fit(x_train, y_train)
print('Alpha : ', ridge_cv.alpha_)
```

```python
ridge_model = Ridge(alpha=1)
```

```python
ridge_model.fit(x_train, y_train)
```

```python
ridge_pred = ridge_model.predict(x_test)
```

```python
ridge_pred
```

```python
ridge_mse = mean_squared_error(y_test, ridge_pred)
ridge_r2 = r2_score(y_test, ridge_pred)
print('MSE: ', ridge_mse)
print('R2 score: ', ridge_r2)
print('cofficients: ', ridge_model.coef_)
```

# ElasticNet

```python
from sklearn.linear_model import ElasticNetCV
elastic_cv = ElasticNetCV(
    l1_ratio = [0.1,0.3,0.5,0.7,0.9,1],
    alphas = [0.001,0.01,0.1,1,10],
    cv = 5,
    random_state = 42
)
elastic_cv.fit(x_train, y_train)
print('Alpha : ', elastic_cv.alpha_)
print('l1_ratio : ', elastic_cv.l1_ratio_)
```

```python
elastic_model = ElasticNet(alpha=0.01, l1_ratio=0.1)
elastic_model.fit(x_train, y_train)
```

```python
elastic_pred = elastic_model.predict(x_test)
```

```python
elastic_pred
```

```python
# Regularization Techniques -> 
#  -> These techniques are use to reduce the overfitting 

# 1. Ridge => 
#  -> it changes the coefficients near to the zero
#  -> It does not change the coefficients to exactly zero 

# 2. Lasso => 
#  -> It changes the coefficients value to the zero 


# 3. Elastic Net => 
#  -> It is a combination of Ridge and Lasso 
#  -> It can make coefficients to zero or it can changes them near to the zero 

# ==================================================================================

# a. MAE
#  -> It stands for Mean Absolut Error
#  -> When we want to treat the errors as same 
#  -> for example there are two errors 
# 	5 and -5 the both treat as same 

# b. MSE
#  -> It stands for mean square error
#  -> it squared the error and then find it's mean 
#  -> It makes the errors differentiable 
#  -> It makes the large errors more large 
#  -> It affects the units
#  -> example => 
# 	2, 10 => 4, 100

# c. RMSE
#  -> It stands for Root mean square error 
#  -> it is the root value of MSE 
#  -> It does not affect the unit

# d. R2 score
#  -> When we wants to know that how much our model is correct 
	
```
