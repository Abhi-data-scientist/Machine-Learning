# Decision Tree model 

```python
# Decision tree model --> CART(Classificatoin And Regression Tree)

# When data is numerical --> branch individual output average --> final output 
# when data is categorical --> branch individual output (100 covid patients) --> 70 yes and 30 no -> final o/p -> 70 yes
# loss - -p_yes*log_yes - p_no*log_no
```

```python
# When data is numerical 
```

```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv('insurance - insurance.csv')
df.head(3)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['sex'] = le.fit_transform(df['sex'])
df['smoker'] = le.fit_transform(df['smoker'])
df['region'] = le.fit_transform(df['region'])
```

```python
x = df.drop(columns=['charges'])
y = df['charges']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dt = DecisionTreeRegressor()
```

```python
dt.fit(x_train, y_train)
```

```python
y_pred = dt.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```
