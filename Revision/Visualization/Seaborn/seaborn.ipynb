```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

| Feature           | Matplotlib                 | Seaborn                                      |
| ----------------- | -------------------------- | -------------------------------------------- |
| Level             | Low-level plotting library | High-level statistical visualization library |
| Ease of use       | More code required         | Simpler, cleaner syntax                      |
| Appearance        | Basic default style        | Attractive default themes                    |
| Statistical plots | Limited                    | Built-in support                             |
| Customization     | Very flexible              | Good, but built on Matplotlib                |
| Best for          | Fine-grained control       | Quick, informative visualizations            |


```python
df = sns.load_dataset("titanic")
df.head()
```

```python
# tells the frequency
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True,
    color="black"
)

plt.title("Age Distribution")
plt.show()
```

```python
# tells the spread of data
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="age",
    fill=True
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare"
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived"
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    size="pclass",
    hue="survived"
)

plt.show()
```

```python
avg_age = df.groupby("pclass")["age"].mean().reset_index()

sns.lineplot(
    data=avg_age,
    x="pclass",
    y="age",
    marker="o"
)

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="sex",
    y="fare"
)

plt.show()
```

```python
sns.violinplot(
    data=df,
    x="sex",
    y="fare"
)
```

```python
# it is the pair of histplot and scatter plot
sns.pairplot(
    df[
        [
            "age",
            "fare",
            "sibsp",
            "parch",
            "survived"
        ]
    ],
    hue="survived"
)
```

```python
corr = df.select_dtypes("number").corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```
