# 📌Part B – Data Visualization 

```python
plt.figure(figsize=(8,5))
sns.histplot(
    df["adr"],
    bins=30, 
    kde=True
)

plt.title("Average Daily Rate (ADR) Distribution")
```

1. Is the distribution normally distributed? 
> No

2. Is it positively or negatively skewed?
> positively skewed

3. Are there any unusually high room prices?
> Yes There are a few extremely high ADR values, as indicated by the long right tail.

#### Question 2 – KDE Plot

```python
sns.kdeplot(
    df, 
    x="lead_time", 
    fill=True,
    color="green"
)
```

1. At approximately what lead time is the booking density highest?
> around 10–20

2. Do most customers book well in advance or close to the check-in date?
> Most customers book close to the check-in date.

#### Question 3 – Scatter Plot

```python
plt.figure(figsize=(8,5))
sns.scatterplot(df, x="lead_time", y="adr", 
                hue="is_canceled", 
               style="hotel", 
               size="total_of_special_requests", alpha=0.7)

plt.title("Relationship between Lead Time and ADR")
plt.xlabel("Lead Time (Days)")
plt.ylabel("ADR (Average Daily Rate)")
plt.legend(title="Legend", bbox_to_anchor=(1.05, 1), loc="upper left")
```

1. Is there any relationship between Lead Time and ADR?
> There is no strong linear relationship between Lead Time and ADR.
> The points are widely scattered, indicating a weak or no correlation.

2. Which type of bookings are canceled more frequently?
> This cannot be concluded from this plot alone.

3. Do customers requesting more special services tend to cancel less?
> This also cannot be confirmed from this scatter plot.

#### Question 4 – Box Plot

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="hotel",
    y="adr",
    palette="Set2"
)

plt.title("ADR Distribution Across Hotel Types")
plt.xlabel("Hotel Type")
plt.ylabel("ADR (Average Daily Rate)")

plt.show()
```

1. Which hotel type has the higher average room price?

City Hotel has a slightly higher room price.

2. Which hotel contains more outliers?

City Hotel contains more outliers, including one extreme ADR outlier.

3. What does the median line represent?

The median line represents the middle (50th percentile) value of ADR, where half the observations lie above it and half lie below it.

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="customer_type",
    y="lead_time",
    hue="is_canceled",
    split=True,          # Split the violin by cancellation status
    palette="Set2"
)

plt.title("Distribution of Lead Time by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Lead Time (Days)")
plt.legend(title="Booking Canceled")

plt.show()
```

1. Which customer type books the earliest?

Transient customers have the lowest lead times and therefore book the earliest.

2. Which customer type shows the highest booking cancellations?

Contract customers show the highest concentration of canceled bookings at higher lead times (based on this violin plot).

3. How is a violin plot different from a box plot?

A violin plot shows both the distribution (density) and summary statistics, while a box plot only shows summary statistics such as the median, quartiles, and outliers.

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="market_segment",
    hue="is_canceled",
    palette="Set2"
)

plt.title("Number of Bookings by Market Segment")
plt.xlabel("Market Segment")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=45)
plt.legend(title="Booking Canceled")

plt.tight_layout()
plt.show()
```

1. Which market segment contributes the highest number of bookings?

Online TA contributes the highest number of bookings.

2. Which market segment has the highest number of cancellations?

Online TA has the highest number of booking cancellations.

```python
plt.figure(figsize=(8,6))

sns.barplot(
    data=df,
    x="meal",
    y="adr",
    palette="magma"      # You can also use Set2, pastel, coolwarm, magma
)

plt.title("Average ADR by Meal Type")
plt.xlabel("Meal Type")
plt.ylabel("Average ADR (Average Daily Rate)")

plt.tight_layout()
plt.show()
```

1. Which meal package has the highest average room price?

HB (Half Board) has the highest average room price (ADR).

2. Why is a bar plot suitable for this analysis instead of a count plot?

A bar plot is suitable because it displays the average ADR for each meal type. A count plot only shows the number of bookings in each category and cannot compare average room prices

```python
# Select only numerical columns
numeric_df = df.select_dtypes(include=["int", "float"])

# Correlation matrix
corr_matrix = numeric_df.corr()

# Plot Heatmap
plt.figure(figsize=(14,10))

sns.heatmap(
    corr_matrix,
    annot=True,          # Display correlation values
    cmap="coolwarm",     # Color palette
    fmt=".2f",           # Show 2 decimal places
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

1. Which two features have the strongest positive correlation?

stays_in_weekend_nights and stays_in_week_nights have the strongest positive correlation (≈ 0.50).

2. Which feature has the strongest negative correlation?

arrival_date_year and arrival_date_week_number have the strongest negative correlation (≈ -0.54).

3. Which numerical features appear to have very little relationship?

Features such as arrival_date_day_of_month, babies, and booking_changes have correlation values close to 0 with most other variables, indicating very little linear relationship.
