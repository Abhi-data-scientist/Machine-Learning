```python
import numpy as np

np.random.seed(42)

# Exponential distribution (right-skewed)
arr = np.random.exponential(scale=2, size=1000)
```

```python
import seaborn as sns
sns.histplot(arr, kde=True)
```

```python
# Log transformation -> When the data range is very high
# Square root -> range is in lower range 6
```

```python
new = np.log1p(arr)
sns.histplot(new, kde=True)
```

```python
new1 = np.sqrt(arr)
sns.histplot(new1, kde=True)
```
