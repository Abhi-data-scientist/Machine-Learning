```python
import numpy as np
import pandas as pd
```

```python
# Handling Missing Values
```

```python
df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 28],
    "Gender": ["Male", "Female", "Male", np.nan, "Male"]
})
```

```python
df["Age"].isnull().sum()
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
```

```python
mode = df["Gender"].mode()[0]
df["Gender"].fillna(mode, inplace=True)
```
