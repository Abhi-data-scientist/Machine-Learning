# Handling Missing Values

```python
df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 28],
    "Gender": ["Male", "Female", "Male", np.nan, "Male"]
})
```

```python
df["Age"].isnull().sum()
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
```

```python
mode = df["Gender"].mode()[0]
df["Gender"].fillna(mode, inplace=True)
```

# Handling Outliers

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
```

```python
df = pd.read_csv("train - train.csv")
df.head(5)
```

```python
Q1 = df["LotArea"].quantile(0.25)
Q3 = df["LotArea"].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR 
upper = Q3+1.5*IQR 
```

```python
# Removing 
```

```python
sns.boxplot(df["LotArea"])
```

```python
new_dataset = df[(df["LotArea"]>lower) & (df["LotArea"]<upper)]
```

```python
sns.boxplot(new_dataset["LotArea"])
```

```python
# Capping
```

```python
new = df["LotArea"].clip(upper=upper, lower=lower)
```

```python
sns.boxplot(new)
```
