```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.compose import ColumnTransformer 
from sklearn.impute import SimpleImputer 
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("Churn_Modelling - Churn_Modelling.csv")
df.sample(5)
```

```python
df.drop(columns=["RowNumber", "CustomerId", "Surname"], inplace=True)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
cat_col = df.select_dtypes(include=str)
```

```python
le = LabelEncoder()
```

```python
for i in cat_col: 
    df[i] = le.fit_transform(df[i])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

# Outliers

```python
# trimming
```

```python
percentile25 = df["Age"].quantile(0.25)
percentile75 = df["Age"].quantile(0.75)
```

```python
IQR = percentile75-percentile25
```

```python
upper_limit = percentile75 + 1.5*IQR 
lower_limit = percentile25 - 1.5*IQR
```

```python
# df = df[(df["Age"]>lower_limit) & (df["Age"]<upper_limit)]
```

```python
df.shape
```

```python
# Capping
```

```python
sns.boxplot(x=df['Age'])
```

```python
df["Age"] = df["Age"].clip(lower_limit, upper_limit)
```

```python
df.shape
```

```python
sns.boxplot(x=df['Age'])
```

```python
x = df.drop(columns=["Exited"])
y = df["Exited"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
x_train1, x_test1, y_train1, y_test1 = train_test_split(x, y, test_size=0.20, random_state=42)
x_train2, x_test2, y_train2, y_test2 = train_test_split(x, y, test_size=0.20, random_state=42)
```
